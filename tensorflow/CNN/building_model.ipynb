```python
import tensorflow as tf
from tensorflow.keras.layers import (
              Conv2D, 
              MaxPool2D, 
              InputLayer, 
              BatchNormalization, 
              Input, 
              Dense, 
              Flatten,
)
from tensorflow import keras 
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Layer
```

### Building A Model 

#### Sequentioal API

```python
seq_model = keras.Sequential([
       InputLayer(shape=(224, 224, 3)),

       Conv2D(filters=8, kernel_size=3, strides=1, padding='valid', activation='relu'),
       BatchNormalization(),
       MaxPool2D(pool_size=2, strides=2),

       Conv2D(filters=16, kernel_size=3, strides=1, padding='valid', activation='relu'),
       BatchNormalization(),
       MaxPool2D(pool_size=2, strides=2),

       Conv2D(filters=32, kernel_size=3, strides=1, padding='valid', activation='relu'),
       BatchNormalization(),
       MaxPool2D(pool_size=2, strides=2),

       Flatten(),

       Dense(1000, activation='relu'),
       BatchNormalization(),

       Dense(100, activation='relu'),
       BatchNormalization(),

       Dense(1, activation='sigmoid'),

], name='sequential_model')

seq_model.summary()
```

#### Functional API

```python
func_input = Input(shape=(224, 224, 3), name='Image Input')

x = Conv2D(filters=8, kernel_size=3, strides=1, padding='valid', activation='relu')(func_input)
x = BatchNormalization()(x)
x = MaxPool2D(pool_size=2, strides=2)(x)

x = Conv2D(filters=16, kernel_size=3, strides=1, padding='valid', activation='relu')(x)
x = BatchNormalization()(x)
x = MaxPool2D(pool_size=2, strides=2)(x)

x = Conv2D(filters=32, kernel_size=3, strides=1, padding='valid', activation='relu')(x)
x = BatchNormalization()(x)
x = MaxPool2D(pool_size=2, strides=2)(x)

x = Flatten()(x)

x = Dense(1000, activation='relu')(x)
x  = BatchNormalization()(x)

x = Dense(100, activation='relu')(x)
x = BatchNormalization()(x)

func_output = Dense(1, activation='sigmoid')(x)

func_model = Model(func_input, func_output, name="Functional_model")

func_model.summary()
```

#### Sequential + Functional API

```python
fe_model = keras.Sequential([
       InputLayer(shape=(224, 224, 3)),
       Conv2D(filters=8, kernel_size=3, strides=1, padding='valid', activation='relu'),
       BatchNormalization(),
       MaxPool2D(pool_size=2, strides=2),

       Conv2D(filters=16, kernel_size=3, strides=1, padding='valid', activation='relu'),
       BatchNormalization(),
       MaxPool2D(pool_size=2, strides=2),

       Conv2D(filters=32, kernel_size=3, strides=1, padding='valid', activation='relu'),
       BatchNormalization(),
       MaxPool2D(pool_size=2, strides=2),
       ], 
       name="Feature_extractor_model")

fe_model.summary()
```

```python
func_input = Input(shape=(224, 224, 3))

x = fe_model(func_input)

x = Flatten()(x)

x = Dense(1000, activation='relu')(x)
x  = BatchNormalization()(x)

x = Dense(100, activation='relu')(x)
x = BatchNormalization()(x)

func_output = Dense(1, activation='sigmoid')(x)

seq_func_model = Model(func_input, func_output, name="Sequential_Functional_model")
seq_func_model.summary()
```

#### SubClassing  --> Custum Layer or Model

```python
class FeatureExtractor(Layer):
    def __init__(self, filters=8, kernel_size=3, padding='valid', strides=1, activation='relu'):
        super(FeatureExtractor, self).__init__()

        self.conv1 = Conv2D(filters=filters, kernel_size=kernel_size, padding=padding, strides=strides, activation=activation)
        self.batch1 = BatchNormalization()
        self.maxpool1 = MaxPool2D(pool_size=2, strides=2*strides)

        self.conv2 = Conv2D(filters=2*filters, kernel_size=kernel_size, padding=padding, strides=strides, activation=activation)
        self.batch2 = BatchNormalization()
        self.maxpool2 = MaxPool2D(pool_size=2, strides=2*strides)

    def call(self, x):
        x = self.conv1(x)
        x = self.batch1(x)
        x = self.maxpool1(x)
        x = self.conv2(x)
        x = self.batch2(x)
        x = self.maxpool2(x)

        return x
    
feature_extractor = FeatureExtractor()
```

```python
class CNNModel(Model):
    def __init__(self, input_shape=(224, 224, 3), filters=8, kernel_size=3, padding='valid', strides=1, activation='relu', dense1_size=1000, dense2_size=100, activation_dense='relu'):
        super(CNNModel, self).__init__()
        
        # self.input = Input(shape=input_shape)
        self.feature_extractor = FeatureExtractor(filters, kernel_size, padding, strides, activation)
        self.flatten = Flatten()
        self.dense1 = Dense(dense1_size, activation=activation_dense)
        self.batch3 = BatchNormalization()
        self.dense2 = Dense(dense2_size, activation=activation_dense)
        self.batch4 = BatchNormalization()
        self.dense3 = Dense(1, activation='sigmoid')

    def call(self, x):
        
        # x = self.input(x)
        x = self.feature_extractor(x)
        x = self.flatten(x)
        x = self.dense1(x)
        x = self.batch3(x)
        x = self.dense2(x)
        x = self.batch4(x)
        x = self.dense3(x)

        return x
    

cnn_model = CNNModel()
cnn_model.summary()
```

```python
class MyDense(Layer):
    def __init__(self, output_shape):
        super(MyDense, self).__init__()
        self.output_shape = output_shape

    
    def build(self, input_shape):
        self.weight = self.add_weight(shape=(input_shape[-1], self.output_shape), initializer='random_normal', trainable=True)
        self.biase = self.add_weight(shape=(self.output_shape,), trainable=True, initializer='zeros')

    def call(self, input_feature):
        return tf.matmul(input_feature, self.weight) + self.biase
```
