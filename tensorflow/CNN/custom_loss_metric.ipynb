##### Custom Loss

```python
def customBCE(factor):
    def func(y_pred, y_true):
        bce = keras.losses.BinaryCrossentropy() 
        return bce(y_pred, y_true) * factor
    return func

# custom class
class CustomBCE(keras.losses.Loss):
    def __init__(self, factor):
        super(CustomBCE, self).__init__()
        self.factor = factor

    def __call__(self, y_pred, y_true):
         bce = keras.losses.BinaryCrossentropy() 
         return bce(y_pred, y_true) * self.factor
```
